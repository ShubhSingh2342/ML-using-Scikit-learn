```python
# import libraries
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
# loading excel file
data = pd.read_excel('iris-train.xlsx')
data.head()
```

```python
# separate features and labels
x = data.iloc[:, :-1] # all columns except last for features
y = data.iloc[:, -1] # last column as features
```

```python
# train the model
model = RandomForestClassifier()
model.fit(x.values, y.values)

datatest = pd.read_excel("iris-test.xlsx")
testx = datatest.iloc[:,:-1]
testy = datatest.iloc[:, -1]
```

```python
# inference
predictions = model.predict(testx.values)
predictions
```

```python
# Accuracy - tells how often the model was right
# accuracy = correct prediction / total predictions = 90/100
actual = testy.values
actual
```

```python
count = 0
for i in range(0, len(actual)):
    if actual[i] == predictions[i]:
        count = count + 1
print((count*100)/len(actual))
```
